**Setup.** Loads PyTorch and our text: the sentences your model will learn from.

```python
import torch

text = "the cat sat on the mat. the dog sat on the log."
```

**Find every different character.** Each letter, the space and the full stop count once, sorted. These are all the characters your model knows.

🤔 *Guess what this prints, then press ▶.*

```python
# Find every different character, sorted
unique_chars = sorted(set(text))
print(unique_chars)
```

**Give each character a number.** Two lookup tables: character → number, and number → character.

```python
# Give each one a number, and the reverse
char_to_number = {}
number_to_char = {}
for number, char in enumerate(unique_chars):
    char_to_number[char] = number
    number_to_char[number] = char
```

**Text → numbers.** `encode` swaps each character for its number. Models only work with numbers.

```python
# Text -> numbers
def encode(some_text):
    numbers = []
    for char in some_text:
        numbers.append(char_to_number[char])
    return numbers
```

**Numbers → text.** `decode` swaps the numbers back, so we can read what the model writes.

🤔 *Guess what this prints, then press ▶.*

```python
# Numbers -> text
def decode(numbers):
    result = ""
    for number in numbers:
        result = result + number_to_char[number]
    return result

print(encode("cat"))

print(decode([4, 11, 6]))
```

**The whole text as a tensor.** Our text, now one row of numbers, ready for the model.

🤔 *Guess what this prints, then press ▶.*

```python
# The whole text as a tensor
print(torch.tensor(encode(text)))
```
